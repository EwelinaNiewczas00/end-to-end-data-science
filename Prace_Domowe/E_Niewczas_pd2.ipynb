# Praca domowa 2: pomiar pamięci i czasu obliczeń w notebooku Jupyter





```python
import sys
import time

lista_1 = list(range(50))
lista_2 = list(range(500))
lista_3 = list(range(5000))
lista_4 = list(range(5000000))


print(f"rozmiar 0-49:" , sys.getsizeof(lista_1), "bajtow")
print(f"rozmiar 0-499:" , sys.getsizeof(lista_2), "bajtow")
print(f"rozmiar 0-4999:" , sys.getsizeof(lista_3), "bajtow")
print(f"rozmiar 0-499999:" , sys.getsizeof(lista_4), "bajtow")
```

```python
#sposób2
start = time.time()

srednia = sum(lista_4) / len(lista_4)

koniec = time.time()
wynik = koniec - start

print(f"Obliczona srednia: {srednia}")
print(f"Czas trwania obliczeń dla 4,99999 milionów liczb: {wynik:.5f} sekund")
```

```python
#wszystkie przedziały
pakiety_list = [lista_1, lista_2, lista_3, lista_4]

for i, lst in enumerate(pakiety_list):
    start = time.time()
    srednia = sum(lst) / len(lst)
    koniec = time.time()
    czas = koniec - start
    print(f"kista {len(lst)} elementów, Średnia: {srednia},  Czas: {czas:.10f} s")
```

```python
#sposób3
n = 5000000

start_2 = time.time()

srednia_szybka = (n - 1) / 2

koniec_2 = time.time()
czas_szybki = koniec_2- start_2

print(f"Obliczona średnia wzorem: {srednia_szybka}")
print(f"Czas trwania nowego sposobu: {czas_szybki:.10f} sekund")
```

```python
#wszystkie przedziały
przedzialy = [50, 500, 5000, 5000000]

for n in przedzialy:
    start = time.time()
    srednia = (n - 1) / 2
    koniec = time.time()
    czas = koniec - start
    print(f"Przedział do {n-1} Średnia {srednia}  Czas: {czas:.8f} s")
```
