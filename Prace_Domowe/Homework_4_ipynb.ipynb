## Zadanie 2 — Kursy walut: konwersje, procenty i formatowanie

Analizujesz prostą wymianę waluty przed krótkim wyjazdem. Część danych liczbowych została zapisana jako tekst i trzeba ją najpierw przekonwertować.

```python
amount_pln = "12750.50"
eur_rate_today = 4.2685
eur_rate_previous = 4.2140
transaction_fee_percent = "1.8"
```

1. Sprawdź typy wszystkich zmiennych wejściowych.
2. Przekonwertuj `amount_pln` i `transaction_fee_percent` na odpowiednie typy liczbowe.
3. Oblicz kwotę w EUR przed prowizją.
4. Oblicz wartość prowizji w EUR oraz kwotę po prowizji.
5. Oblicz procentową zmianę kursu względem `eur_rate_previous`.
6. Oblicz bezwzględną różnicę między kursem dzisiejszym i poprzednim z użyciem `abs()`.
7. Utwórz co najmniej dwa wyniki logiczne, na przykład czy kurs wzrósł oraz czy zmiana kursu przekracza 1%.
8. Wypisz czytelny komunikat z użyciem `print()` i f-stringów.

```python
print(f"Zadanie 2.1")
print(type(amount_pln))
print(type(eur_rate_today))
print(type(eur_rate_previous))
print(type(transaction_fee_percent))
print(f"Zadanie 2.2")
amount_pln_float = float(amount_pln)
transaction_fee_percent_float = float(transaction_fee_percent)
print(f"Zadanie 2.3")
euro = (amount_pln_float / eur_rate_today)
print(f"{euro:.2f} euro")
print(f"Zadanie 2.4")
prowizja = euro * (transaction_fee_percent_float / 100)
euro_po_prowizji = euro - prowizja
print(f"{euro_po_prowizji:.2f} euro")
print(f"Zadanie 2.5")
zmiana_kursu_procent = ((eur_rate_today - eur_rate_previous) / eur_rate_previous) * 100
print(f"Zmiana procentowa kursu: {zmiana_kursu_procent:+.2f}%")
print(f"Zadanie 2.6")
roznica_kursow = abs(eur_rate_today - eur_rate_previous)
print(f"Bezwzględna różnica: {roznica_kursow:.4f}")
print(f"Zadanie 2.7")
czy_kurs_wzrosl = eur_rate_today > eur_rate_previous
czy_duza_zmiana = abs(zmiana_kursu_procent) > 1.0
print(f"Czy kurs wzrósł? {czy_kurs_wzrosl}")
print(f"Czy zmiana jest duża (>1%)? {czy_duza_zmiana}")
print("Zadanie 2.8")
print(f"--- PODSUMOWANIE TRANSAKCJI ---")
print(f"Wymieniana kwota: {amount_pln_float:,.2f} PLN")
print(f"Kurs dzisiejszy:  {eur_rate_today:.4f} PLN | Kurs poprzedni: {eur_rate_previous:.4f} PLN")
print(f"Czy kurs wzrósł?  {czy_kurs_wzrosl}")
print(f"Czy zmiana > 1%?  {czy_duza_zmiana}")
print(f"Ostateczna kwota w EUR po odliczeniu prowizji: {euro_po_prowizji:.2f} EUR")
```

## Zadanie 3 — Pomiary odległości: biblioteka math i dokładność obliczeń

Masz współrzędne dwóch punktów na planie pomieszczenia. Chcesz policzyć odległość między nimi i sprawdzić, czy zgadza się z wartością referencyjną.

```python
import math

x1 = 2.5
y1 = 1.0
x2 = 11.5
y2 = 7.0
expected_distance = 10.82
```

1. Oblicz różnicę współrzędnych `dx` i `dy`.
2. Oblicz odległość między punktami z użyciem `math.sqrt()`.
3. Zaokrąglij wynik do dwóch miejsc po przecinku.
4. Sprawdź za pomocą `math.isclose()`, czy obliczona odległość jest bliska `expected_distance`.
5. Wypisz wynik logiczny bez użycia `if`.
6. Przygotuj krótki raport tekstowy z użyciem f-stringów i `print()`.

```python
print(f"Zadanie 3.1")
dx = x2 - x1
dy = y2 - y1
print(f"dx = {dx}, dy = {dy}")

print(f"Zadanie 3.2 i 3.3")
odleglosc = math.sqrt(dx**2 + dy**2)
odleglosc_zaokraglona = round(odleglosc, 2)
print(f"Odległość (dokładna): {odleglosc}")
print(f"Odległość (zaokrąglona): {odleglosc_zaokraglona}")

print(f"Zadanie 3.4 i 3.5")
czy_blisko = math.isclose(odleglosc_zaokraglona, expected_distance, rel_tol=0.01)
print(f"Czy wynik jest bliski oczekiwanemu? {czy_blisko}")

print(f"Zadanie 3.6")
print(f"--- RAPORT ---")
print(f"Punkt A: ({x1}, {y1}), Punkt B: ({x2}, {y2})")
print(f"Wyliczona odległość wynosi: {odleglosc_zaokraglona}")
print(f"Oczekiwana odległość:{expected_distance}")
print(f"Zgodność z oczekiwaniem: {czy_blisko}")
```

## Zadanie 4 — Organizacja próbek: dzielenie całkowite, reszta i zaokrąglenia math

Planujesz logistykę dla serii próbek laboratoryjnych. Trzeba policzyć pudełka, czas pracy i prosty raport organizacyjny.

```python
import math

n_samples = 358
box_capacity = 24
n_analysts = 5
minutes_per_sample = 7
```

1. Oblicz liczbę pełnych pudełek z użyciem `//`.
2. Oblicz liczbę próbek pozostających poza pełnymi pudełkami z użyciem `%`.
3. Oblicz minimalną liczbę pudełek potrzebnych na wszystkie próbki z użyciem `math.ceil()`.
4. Oblicz łączny czas pracy w minutach dla wszystkich próbek.
5. Załóż, że analitycy dzielą pracę po równo. Oblicz przybliżoną liczbę minut pracy na jednego analityka.
6. Oblicz pełne godziny pracy z użyciem `math.floor()` oraz liczbę pozostałych minut.
7. Przygotuj raport tekstowy z użyciem `print()` i f-stringów.

```python
print(f"Zadanie 4.1")
pelne_pudelka = n_samples // box_capacity
print(f"Liczba pełnych pudełek: {pelne_pudelka}")

print(f"Zadanie 4.2")
pozostale_probki = n_samples % box_capacity
print(f"Próbki poza pełnymi pudełkami: {pozostale_probki}")

print(f"Zadanie 4.3")
wszystkie_pudelka = math.ceil(n_samples / box_capacity)
print(f"Minimalna liczba pudełek ogółem: {wszystkie_pudelka}")

print(f"Zadanie 4.4")
calkowity_czas_minuty = n_samples * minutes_per_sample
print(f"Łączny czas pracy: {calkowity_czas_minuty} minut")

print(f"Zadanie 4.5")
czas_na_analityka = calkowity_czas_minuty / n_analysts
print(f"Czas na jednego analityka: {czas_na_analityka} minut")

print(f"Zadanie 4.6")
pelne_godziny = math.floor(calkowity_czas_minuty / 60)
pozostale_minuty = calkowity_czas_minuty % 60
print(f"Pełne godziny: {pelne_godziny}, pozostałe minuty: {pozostale_minuty}")

print(f"Zadanie 4.7")
print(f"--- RAPORT ---")
print(f"Liczba wszystkich próbek: {n_samples}")
print(f"Wymagana liczba pudełek:  {wszystkie_pudelka} (w tym pełnych: {pelne_pudelka})")
print(f"Czas pracy na analityka:  {czas_na_analityka:.1f} minut")
print(f"Łączny czas laboratorium: {pelne_godziny} godz. oraz {pozostale_minuty} min.")
```

```python
print(f"Zadanie 5.1")
clean_report_name = raw_report_name.strip()
clean_column_name = raw_column_name.strip()
print(f"Oczyszczony raport: '{clean_report_name}'")
print(f"Oczyszczona kolumna: '{clean_column_name}'")

print(f"Zadanie 5.2")
lower_report_name = clean_report_name.lower()
print(f"Małe litery: '{lower_report_name}'")

print(f"Zadanie 5.3")
replaced_report_name = lower_report_name.replace("-", "_").replace(" ", "_")
print(f"Po zamianie znaków: '{replaced_report_name}'")

print(f"Zadanie 5.4 i 5.5")
is_csv = replaced_report_name.endswith(".csv")
starts_with_report = replaced_report_name.startswith("report")
print(f"Czy kończy się na .csv? {is_csv}")
print(f"Czy zaczyna się od 'report'? {starts_with_report}")

print(f"Zadanie 5.6")
name_without_ext = replaced_report_name.replace(".csv", "")
parts = name_without_ext.split("_")
region = parts[1]
rok = parts[2]
miesiac = parts[3]
temat = parts[4]
print(f"Region: {region}, Rok: {rok}, Miesiąc: {miesiac}, Temat: {temat}")

print(f"Zadanie 5.7")
rok_miesiac_slice = clean_report_name[19:26]
print(f"Wycięty fragment (slicing): {rok_miesiac_slice}")

print(f"Zadanie 5.8")
column_snake = clean_column_name.lower().replace(" ", "_")
column_title = clean_column_name.title()
print(f"Kolumna snake_case: '{column_snake}'")
print(f"Kolumna Title Case: '{column_title}'")

print(f"Zadanie 5.9")
elementy_nazwy = ["clean", region, rok, miesiac, temat + ".csv"]
nowa_nazwa_pliku = "_".join(elementy_nazwy)
print(f"Nowa nazwa pliku (join): {nowa_nazwa_pliku}")

print(f"Zadanie 5.10")
dlugosc_nazwy = len(nowa_nazwa_pliku)
print(f"--- RAPORT ---")
print(f"Nowa nazwa:  {nowa_nazwa_pliku}")
print(f"Długość nowej nazwy wynosi: {dlugosc_nazwy} znaków.")
```
