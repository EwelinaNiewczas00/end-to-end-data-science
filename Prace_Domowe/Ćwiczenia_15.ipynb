```python
import numpy as np
import pandas as pd

#wektor jako tablica 1d
wektor_a = np.array([1,2,3,4])
print(f"Wektor a: {wektor_a}")
print(f"Kształt: {wektor_a.shape}")
print(f"Typ: {type(wektor_a)}")

#wektor kolumnowy
wektor_kolumna = np.array([[1], [2], [3], [4]])
print("\nWektor kolumnowy")
print(wektor_kolumna)
print(f"\nKształt: {wektor_kolumna.shape}")

#wektor wierszowy
wektor_wiersz = np.array([[1,2,3,4]])
print("\nWektor wierszowy:")
print(wektor_wiersz)
print(f"Kształt: {wektor_wiersz.shape}")

#reshape między formatami
wektor_reshaped = wektor_a.reshape(-1,1)  #-1 = automatyczny wymiar
print("\nReshape 1d - kolumnowy", wektor_reshaped.shape)

```

```python
from sklearn.datasets import load_iris
# Wczytanie Iris dataset
iris = load_iris()

X = iris.data # Macierz cech (150, 4)
y = iris.target # Wektor etykiet (150,)

print("Kształt macierzy cech:", X.shape) # 150 obserwacji, 4 cechy
print("Kształt wektora etykiet:", y.shape) # 150 etykiet

# Pierwszy kwiat (obserwacja) jako wektor cech
kwiat_0 = X[0]
print("\nPierwszy kwiat (wektor cech):", kwiat_0)
print("Cechy: sepal_length, sepal_width, petal_length, petal_width")
# Interpretacja: punkt w przestrzeni 4D
print(f"\nKwiat 0 to punkt w R^4:")
print(f" sepal_length = {kwiat_0[0]} cm")
print(f" sepal_width = {kwiat_0[1]} cm")
print(f" petal_length = {kwiat_0[2]} cm")
print(f" petal_width = {kwiat_0[3]} cm")
# Pojedyncza cecha dla wszystkich kwiatów to również wektor
sepal_lengths = X[:, 0] # Wektor długości działek (150,)
print(f"\nDługości działek (wektor 150 wartości):")
print(f" Średnia: {sepal_lengths.mean():.2f}")
print(f" Min: {sepal_lengths.min():.2f}, Max: {sepal_lengths.max():.2f}")
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

suma = a + b
print(f"Suma wektorów: {suma}")

roznica = b - a
print(f"Roznica wektorow: {roznica}")

c = a *2
print(f"Mnozenie przez skalar: {c}")

iloczyn = a *b
print(f"iloczyn wektorów: {iloczyn}")

iloraz = b/a
print(f"Iloraz wektorów: {iloraz}")


#normalizacja cech

wektorek = np.array([10,20,30,40,50])
min_val = wektorek.min()
max_val = wektorek.max()

#min-max normalizacja (x-min)/(max-min)
normalized = (wektorek-min_val)/(max_val-min_val)
print(f"Cecha oryginalna: {wektorek}")
print(f"Cecha znormalizowana: {normalized}")
```
